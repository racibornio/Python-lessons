```python
import enum
from dotenv import dotenv_values
from openai import OpenAI
import instructor
from pydantic import BaseModel
```

```python
env = dotenv_values("../../.env")
openai_client = OpenAI(api_key=env["OPENAI_API_KEY"])
```

```python
instructor_openai_client = instructor.from_openai(openai_client)
```

```python
class ClassificationCategory(str, enum.Enum):
    SPAM = "spam"
    NOT_SPAM = "not_spam"

class ClassificationResult(BaseModel):
    category: ClassificationCategory
```

```python
text = "IV 2022 - zużycie energii"

res = instructor_openai_client.chat.completions.create(
    model="gpt-4o",
    response_model=ClassificationResult,
    messages=[
        {
            "role": "user",
            "content": f"Klasyfiku tekst: {text}"
        }
    ]
)
```

```python
f"Uznajemy wiadomość za {res.category.value}"
```

```python
def classify_text(text, classification_category):
    class ClassificationResult(BaseModel):
        category: classification_category

    res = instructor_openai_client.chat.completions.create(
        model="gpt-4o-mini",
        response_model=ClassificationResult,
        messages=[
            {
                "role": "user",
                "content": f"Klasyfikujemy wiadomość jako {text}"
            }
        ]
    )

    return res.category.value
```

```python
class ClassificationClass(str, enum.Enum):
    SPAM = "spam"
    NOT_SPAM = "not_spam"

print(classify_text(
    text="Cześć, nie znamy się, ale znam sposób, dzięki ktoremu będziesz super bogaty! Kliknij w link, a dowiesz się więcej!",
    classification_category=ClassificationClass
))

print(classify_text(
    text="Cześć, kiedy jedziemy na rower?",
    classification_category=ClassificationClass
))
```

```python
class ClassificationClass(str, enum.Enum):
    TECH_ISSUE = "tech_issue",
    BILLING = "billing",
    GENERAL_QUERY = "general_query"

print(classify_text(
    text="Wydaje mi się, że ma problem z płatnościami - co zrobić?",
    classification_category=ClassificationClass
))

print(classify_text(
    text="Gdzie znajdę informacje odnośnie kolejnych kursów?",
    classification_category=ClassificationClass
))

print(classify_text(
    text="Moje konto zostało zablokowane - co mam zrobić?",
    classification_category=ClassificationClass
))

print(classify_text(
    text="Fajne te Wasze funkcje - dzięki!",
    classification_category=ClassificationClass
))
```
