```python
from dotenv import dotenv_values
from openai import OpenAI
from IPython.display import Audio
```

```python
env = dotenv_values("../../.env")
openai_client = OpenAI(api_key=env["OPENAI_API_KEY"])
```

```python
response = openai_client.audio.speech.create(
    model="tts-1",
    voice="alloy",
    response_format="mp3",
    input="""
kolejne zadanie programistyczne
"""
)
```

```python
response.write_to_file("tts_alloy_1.mp3")
```

```python
Audio("tts_alloy_1.mp3")
```

```python
def generate_speech(text, voice, output_audio_path):
    response = openai_client.audio.speech.create(
        model="tts-1",
        voice=voice,
        response_format="mp3",
        input=text
    )

    response.write_to_file(output_audio_path)
    return Audio(output_audio_path)
```

```python
generate_speech(
    text="""
Oto przykładowy tekst czytany przez model AI.
""",
voice="onyx",
output_audio_path="nagranie_z_modelu.mp3"
)
```
