```python
from dotenv import dotenv_values
from openai import OpenAI
import instructor
from pydantic import BaseModel
```

```python
env = dotenv_values("../../.env")
openai_client = OpenAI(api_key=env["OPENAI_API_KEY"])
```

```python
instructor_openai_client = instructor.from_openai(openai_client)
```

```python
text = """
Nazywam się Patryk i pracę w IT rozpocząłem w roku 2007 jako tester manualny w firmie Teta. Po trzech miesiącach zmieniłem pracę i pracowałem jako IT specialist w supermarkecie E.Leclerc. Po pięciu miesiącach w 2008 roku pracowałem jako tester w AIG banku, a od września 2008 jako tester w polskiej firmie Insert. Po 3 latach od września 2011 przeszedłem jako tester do Atosa. Po półtora roku opuściłem ich i zaczałem w styczniu 2013 jako tester w Volvo, ale już w październiku zacząłem jako tester w BL Stream. Po kilku latach firma połączyła się z SMT i powstało intive - tam po 6-ciu latach pracy jako QA, w listopadzie 2019 zacząłem pracować jako analityk biznesowy. Robiłem to do roku 2025-go a potem zacząłem pracować jako product owner w funduszach inwestycyjnych - do lipca 2026. Potem do końca października nie miałem pracy, a od listopada zaczynam w firmie Sphera jako product owner. W 2017 roku przeprowadziłem pierwsze szkolenie na certyfikację ISTQB CTFL, a do tej pory łącznie szkoleń przeprowadziłem około 90-ciu.
"""
```

```python
class Person(BaseModel):
    name: str
    age: int
    profession: str
    years_of_experience: int
    salary: int
    is_happy: bool

res = instructor_openai_client.chat.completions.create(
    model="gpt-4o-mini",
    temperature=0,
    response_model=Person,
    messages=[
        {
            "role": "user",
            "content": text
        }
    ]
)

res.model_dump()
```

```python
class ShortStory(BaseModel):
    title: str
    cover_summary: str
    chapters_overview: list[str]

res = instructor_openai_client.chat.completions.create(
    model="gpt-4o-mini",
    temperature=0,
    response_model=ShortStory,
    messages=[
        {
            "role": "user",
            "content": """
Zaproponuj opis stylu życia kierowcy ciężarówki, który spędza pięć dni tygodniowo na trasach międzynarodowych.
"""
        }
    ]
)

res.model_dump()
```

```python
def retrieve_structure(text, response_model):
    res = instructor_openai_client.chat.completions.create(
        model="gpt-4o-mini",
        temperature=0,
        response_model=response_model,
        messages=[
            {
                "role": "user",
                "content": text
            }
        ]
    )
    return res.model_dump()
```

```python
class Person(BaseModel):
    name: str
    age: int
    profession: str
    years_of_experience: int
    salary: int
    is_happy: bool

retrieve_structure(
    text="""
Jestem Patryk, mam 44 lata i pracuję w IT od 19-tu lat. Teraz jestem w okresie po zakończeniu jednej pracy a rozpoczęciem drugiej i kończę kurs z Pythona i AI-a.
""",
response_model=Person
)
```
