```python
from dotenv import dotenv_values
from IPython.display import Markdown
from openai import OpenAI
```

```python
env = dotenv_values("../../.env")
openai_client = OpenAI(api_key=env["OPENAI_API_KEY"])
```

```python
res = openai_client.chat.completions.create(
    model="gpt-4o-mini",
    temperature=0,
    messages=[
        {
            "role": "user",
            "content": """
                Daj mi krótki opis tego jak pogoda zmienia się przy przejściu z późnego lata we wczesną jesień w Polsce.
            """
        }
    ]
)
```

```python
Markdown(res.choices[0].message.content)
```

```python
def generate_text(prompt):
    res = openai_client.chat.completions.create(
        model="gpt-4o-mini",
        temperature=0,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return res.choices[0].message.content
```

```python
generate_text("Wymyśl kilka propozycji na utworzenie własnych projektów wokół AI.")
```
