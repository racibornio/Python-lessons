# Eksplorujemy modele `data-to-value`

## Czym są modele `data-to-value`?

* Modele `data-to-value` to modele, które przewidują wartość jakiejś istotnej zmiennej na podstawie danych. 
* Modele te trenujemy w sposób nadzorowany, co oznacza, że potrzebujemy danych treningowych, w których dla każdego przykładu mamy zarówno dane jak i wartość zmiennej, którą chcemy przewidzieć.
* Modele te należy do rodziny tzw. regresji, czyli przewidywania wartości liczbowych (ang. regression models, regressors).
* Modele te w pełni działają na naszych zasobach, nic nas dodatkowo nie kosztują.
* Możemy spokojnie używać tych modeli z wewnętrznymi danymi, ponieważ nie przechodzą one nigdzie poza nasze środowisko.
* Oczywiście same modele powinny być zabezpieczone przed nieautoryzowanym dostępem.

```python
from pycaret.regression import load_model, predict_model
import pandas as pd
```

## Jak używać modelu `regresji`?

```python
# regression_pipeline został wytrenowany na danych
# z dwoma kolumami `x` i `y` zawierającymi liczby całkowite
model = load_model("regression_pipeline")
```

```python
df = pd.DataFrame([
    {"x": 10},
    {"x": 30},
    {"x": 80},
])
predict_model(model, data=df)
```

## Zamykamy model w funkcji

```python
def predict_value(df, predicted_column_name="predicted"):
    model = load_model("regression_pipeline")
    df = predict_model(model, data=df)

    df = df.rename(columns={"prediction_label": predicted_column_name})
    return df
```

```python
predict_value(
    df=pd.DataFrame([
        {"x": 2},
        {"x": 60},
        {"x": 211},
    ]),
    predicted_column_name="predicted_y"
)
```

## Pomysły `gpt-4o` na użycie modelu

### 1. Przewidywanie Cen Nieruchomości
Model regresji może analizować dane dotyczące nieruchomości, takie jak lokalizacja, metraż, liczba pokoi, rok budowy, i przewidywać wartość rynkową domu lub mieszkania. To jest niezwykle przydatne dla agencji nieruchomości i inwestorów.

### 2. Prognozowanie Sprzedaży Produktów
Model AI może przewidywać przyszłą sprzedaż produktów na podstawie historycznych danych sprzedażowych, sezonowości, trendów rynkowych i kampanii marketingowych. To pomaga firmom w planowaniu produkcji i zarządzaniu zapasami.

### 3. Prognozowanie Zużycia Energii
Modele regresji mogą przewidywać zużycie energii elektrycznej w budynkach na podstawie danych historycznych, takich jak temperatury, godziny szczytu, liczba mieszkańców, co jest użyteczne dla firm energetycznych i zarządców budynków.

### 4. Przewidywanie Wynagrodzeń na Rynku Pracy
AI może analizować dane demograficzne, wykształcenie, doświadczenie zawodowe, lokalizację i przewidywać oczekiwane wynagrodzenia dla różnych zawodów i stanowisk. To może wspierać zarówno pracodawców, jak i kandydatów w negocjacjach płacowych.

### 5. Ocena Ryzyka Kredytowego
Model regresji może przewidywać ryzyko niewypłacalności klientów na podstawie ich historii kredytowej, dochodów, zobowiązań finansowych i innych zmiennych, co jest kluczowe dla banków i instytucji finansowych.

### 6. Prognozowanie Czasu Dostawy
Firmy logistyczne mogą używać modeli regresji do przewidywania czasu dostawy przesyłek na podstawie czynników takich jak odległość, warunki drogowe, pogoda, oraz obciążenie sieci logistycznej.

### 7. Prognozowanie Zysków Firmy
AI może analizować dane finansowe firmy, takie jak przychody, koszty operacyjne, wydatki na marketing, aby przewidywać przyszłe zyski lub straty. To może być kluczowe dla podejmowania decyzji strategicznych.

### 8. Przewidywanie Wyników Egzaminów Uczniów
W edukacji, modele regresji mogą przewidywać wyniki egzaminów na podstawie danych dotyczących frekwencji, ocen z testów próbnych, zaangażowania w zajęcia, co może pomóc nauczycielom w identyfikowaniu uczniów wymagających dodatkowego wsparcia.

### 9. Prognozowanie Popytu na Produkty Sezonowe
AI może przewidywać popyt na produkty sezonowe (np. sprzęt narciarski, stroje kąpielowe) na podstawie danych historycznych, prognoz pogodowych, trendów rynkowych, co pomaga firmom lepiej planować zapasy i kampanie marketingowe.

### 10. Przewidywanie Ceny Akcji na Giełdzie
Modele regresji mogą analizować dane rynkowe, historyczne ceny akcji, wskaźniki ekonomiczne, wydarzenia makroekonomiczne, aby przewidywać przyszłe ceny akcji. Jest to użyteczne dla inwestorów i analityków finansowych.
