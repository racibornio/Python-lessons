# Eksplorujemy modele `data-to-category`

## Czym są modele `data-to-category`?

* Modele `data-to-category` to modele, które przewidują kategorię na podstawie danych. 
* Czasami zamiast słowa kategoria używamy słowa etykieta lub klasa.
* Modele te trenujemy w sposób nadzorowany, co oznacza, że potrzebujemy danych treningowych, w których dla każdego przykładu mamy zarówno dane jak i etykietę.
* Modele te należą do rodziny tzw. klasyfikatorów (ang. classifiers).
* Tworzymy je w celu przewidywania kategorii dla nowych danych, które nie były wcześniej obserwowane.
* Modele te w pełni działają na naszych zasobach, nic nas dodatkowo nie kosztują.
* Możemy spokojnie używać tych modeli z wewnętrznymi danymi, ponieważ nie przechodzą one nigdzie poza nasze środowisko.
* Oczywiście same modele powinny być zabezpieczone przed nieautoryzowanym dostępem.

```python
from pycaret.classification import load_model, predict_model
import pandas as pd
```

## Jak używać modelu `klasyfikacji`?

```python
# classification_pipeline został wytrenowany na danych
# z dwoma kolumami `x` i `y` zawierającymi liczby całkowite
model = load_model("classification_pipeline")
```

```python
df = pd.DataFrame([
    {"x": 2, "y": 8},
    {"x": 3, "y": 2},
    {"x": 1, "y": 1},
])
predict_model(model, data=df)
```

## Zamykamy model w funkcji

```python
def classify_data(df, predicted_column_name="kategoria"):
    model = load_model("classification_pipeline")
    df = predict_model(model, data=df)
    df = df.rename(columns={"prediction_label": predicted_column_name})

    return df
```

```python
classify_data(pd.DataFrame([
    {"x": 2, "y": 8},
    {"x": 6, "y": 2},
    {"x": 11, "y": 11},
]))
```
