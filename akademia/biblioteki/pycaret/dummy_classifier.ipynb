```python
from pycaret.classification import setup, create_model, predict_model
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = []
for i in range(1000):
    if i < 300:
        data.append({
            "x" : i,
            "y" : i + 1,
            "label" : "A"
        })
    else:
        data.append({
            "x" : i,
            "y" : i + 1,
            "label" : "B"
        })

df = pd.DataFrame(data)
df.sample(10)
```

```python
sns.displot(df, x="label")
```

```python
setup(data=df, target="label", session_id=123, n_jobs=1)
```

```python
model = create_model("dummy")
```

```python
with_predictions_df = predict_model(model, data=df)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

sns.histplot(with_predictions_df, x="label", ax=axes[0])
axes[0].set_title("Distribution of Labels")
sns.histplot(with_predictions_df, x="prediction_label", ax=axes[1])
axes[1].set_title("Distribution of Prediction labels")
```
