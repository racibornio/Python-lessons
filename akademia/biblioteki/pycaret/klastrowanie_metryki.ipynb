```python
from pycaret.clustering import setup, create_model, plot_model, pull
from pycaret.datasets import get_data
import pandas as pd
import random
import seaborn as sns
```

```python
data = []

for _ in range(100):
    data.append({
        "x": random.randint(0, 100),
        "y": random.randint(0, 100),
        "label": "A"
    })

    data.append({
        "x": random.randint(200, 300),
        "y": random.randint(0, 100),
        "label": "B"
    })

    data.append({
        "x": random.randint(0, 100),
        "y": random.randint(200, 300),
        "label": "C"
    })

    data.append({
        "x": random.randint(200, 300),
        "y": random.randint(200, 300),
        "label": "D"
    })

fake_df = pd.DataFrame(data)
```

```python
sns.scatterplot(data=fake_df, x="x", y="y", hue="label")
```

```python
setup(fake_df, session_id=123, n_jobs=1)
```

```python
fake_model = create_model("kmeans", ground_truth="label")
```

```python
plot_model(fake_model, plot="cluster")
```

```python
fake_metrics_df = pull()
```

```python
pokemon_df = get_data("pokemon", verbose=False)
```

```python
pokemon_df.sample(10)
```

```python
setup(data=pokemon_df, session_id=123, n_jobs=1)
```

```python
model = create_model("kmeans", num_clusters=6)
```

```python
pokemon_metrics_df = pull()
```

```python
plot_model(model, plot="elbow")
```
