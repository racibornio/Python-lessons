```python
from pycaret.regression import setup, create_model, predict_model, get_config
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import random
```

```python
data = []
for x in range(100):
    e = random.randint(0, 40)
    data.append({
        'x' : x,
        'y' : 2 * x + 1 + e
    })

df = pd.DataFrame(data)
```

```python
sns.scatterplot(data=df, x='x', y='y')
```

```python
setup(data=df, target="y", session_id=123, n_jobs=1)
```

```python
model = create_model("dummy")
```

```python
with_predictions_df = predict_model(model, data=df)
```
