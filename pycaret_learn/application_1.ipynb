```python
import pandas as pd
```

```python
from pycaret.clustering import setup, create_model, assign_model, plot_model, save_model, load_model, predict_model
```

```python
df = pd.read_csv('atc_pre_departure_delays_2025.csv', sep=',')
df.head()
```

```python
len(df)
```

```python
s = setup(df, session_id=111)
```

```python
s.dataset.head()
```

```python
s.dataset_transformed.head()
```

```python
kmeans = create_model('kmeans', num_clusters=20)
```

```python
df_with_clusters = assign_model(kmeans)
df_with_clusters
```

```python
df_with_clusters["Cluster"].value_counts()
```

```python
plot_model(kmeans, plot='cluster')
```

```python
save_model(kmeans, 'atc_pre_departure_delays_2025_pipeline', verbose=False)
```

```python
kmeans_pipeline = load_model('atc_pre_departure_delays_2025_pipeline')
kmeans_pipeline
```
